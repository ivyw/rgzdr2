```python
base_path = Path('/home/matthew/native-repos/rgzdr2/data/')
processed_subjects_path = base_path / 'radio_subjects_subset_processed.json'
```

```python
cache = Path('/home/matthew/native-repos/rgzdr2/data/cache')
```

```python
with open(processed_subjects_path) as f:
    subjects_json = json.load(f)
```

```python
subject = subjects.Subject.from_json(subjects_json[0])
```

```python
np.random.seed(0)
random_ordering = np.arange(len(subjects_json))
np.random.shuffle(random_ordering)
```

```python
for i in range(100):
    i = random_ordering[i]
    subject = subjects.Subject.from_json(subjects_json[i])
    ax = plot_subject.plot_single_subject(subject, cache, ax=None)
    plt.show()
    plt.close(ax.get_figure())
    print(subject.zid)
    for j, bbox in enumerate(subject.bboxes.values()):
        print(f'Box {j}')
        print('\n'.join(f'\t{f}' for f in bbox))
```
